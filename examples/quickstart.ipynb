### Step 3: Interactive Data Discovery
You don't need to know variable names or parameter names in advance. Use discovery commands to see all available schemas directly in your notebook:

```python
# 1. Discover all filterable patient profile variables and their valid operators
biom.fields()
```

```python
# 2. List all available clinical datasets
biom.datasets()
```

```python
# 3. Inspect variables in a specific dataset (by ID or name)
biom.variables(1)
```

```python
# 4. Search variables across all studies
biom.search_variables('glucose')
```

### Step 4: Type-Aware Filtering & Retrieval
Retrieve filtered rows directly into a `pandas.DataFrame`:

```python
# Style A: Fluent method chaining
df = (
    biom.dataset(1)
    .filter(age__gte=18, age__lte=65)
    .filter(gender='Female')
    .sort_by('age', ascending=True)
    .to_dataframe()
)
df.head()
```

```python
# Style B: Declarative F expressions
from biom import F

df = biom.query(
    dataset=1,
    filters=[
        F('age') >= 18,
        F('gender') == 'Female',
    ],
).to_dataframe()
df.head()
```

### Step 5: Premade Data Science Tools
`biom-kit` comes with built-in tools for data cleaning, statistical analysis, and plotting.

```python
# 1. Clean dataset (automatically coerces numbers, parses dates, trims whitespace)
clean_df = biom.tools.clean_dataset(df)

# 2. Data Health & Missingness Report
biom.tools.summarize_health(clean_df)
```

```python
# 3. Statistical summary (mean, median, IQR, skewness)
biom.tools.summarize(clean_df)
```

```python
# 4. Correlation Matrix
biom.tools.correlation_matrix(clean_df)
```

```python
# 5. Cohort Aggregations
biom.tools.group_summary(clean_df, group_by='gender')
```

### Step 6: Publication-Quality Plotting

```python
# Distribution Histogram
biom.tools.plot_distribution(clean_df, column='age', hue='gender')
```

```python
# Correlation Heatmap
biom.tools.plot_correlation_matrix(clean_df)
```

```python
# Missing Data Visualization
biom.tools.plot_missingness(clean_df)
```
